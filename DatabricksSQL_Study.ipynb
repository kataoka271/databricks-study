```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import regexp_extract

# Sparkセッションの作成
spark = SparkSession.builder.appName("RegexExample").getOrCreate()

# サンプルデータの作成
data = [("john.doe@example.com",), ("jane.smith@sample.org",)]
columns = ["email"]

# データフレームの作成
df = spark.createDataFrame(data, columns)

# 正規表現でユーザー名とドメインを抽出
user_pattern = r"(^([^@]+))@"
domain_pattern = r"@([^@]+)$"

df = df.withColumn("username", regexp_extract("email", user_pattern, 2))
df = df.withColumn("domain", regexp_extract("email", domain_pattern, 1))

# 結果表示
display(df)
```

```python
import pyspark.sql.functions as F
import pyspark.sql.types as T

@F.udf(T.ArrayType(T.ArrayType(T.StringType())))
def split_phone_number(phone_number):
    a = phone_number.split("-")
    return [a, a]

df = [d["parts"] for d in spark.sql("select * from samples.bakehouse.sales_customers").select("phone_number").withColumn(
    "parts", split_phone_number(F.col("phone_number"))
).select("parts").collect()]

df[0]
```
